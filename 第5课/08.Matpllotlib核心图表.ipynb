```python
from matplotlib.axes import Axes
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#创建子图
# figure: 画布对象；axes: 子图数组
figure,axes=plt.subplots(1,2,figsize=(16,5),dpi=100)

# 图一：柱状图（世界石油储备）
countries=['中国','美国','俄罗斯','印度','加拿大','法国','英国','德国','意大利','日本'] #国家列表
values=[35,23,18,21,56,78,51,12,18,31]

axes1:Axes=axes[0]
axes1.bar(countries,values,color='orange',width=0.5) #柱状图，width: 柱宽度，color: 柱颜色
axes1.set_title('世界石油储备',fontsize=20)#设置标题
axes1.set_xlabel('国家',fontsize=14)#设置x轴标签
axes1.set_ylabel('石油储备（万桶）',fontsize=14)#设置y轴标签
axes1.grid(linestyle='--',alpha=0.5)#设置网格线，linestyle: 线条样式，alpha: 透明度
#axes1.set_xticklabels(countries,rotation=45,fontsize=10)
#axes1.set_yticklabels(values,fontsize=10)

#图二：饼状图（世界人口）----->擅长比例分析
countries2=['中国','美国','俄罗斯','印度','加拿大','法国','英国','德国','意大利','日本']
values2=[14.3,27.8,16.7,12.6,38.3,23.2,20.1,11.9,6.2,10.6]
axes2:Axes=axes[1]
axes2.pie(values2,labels=countries2,autopct='%1.1f%%',startangle=90)#autopct: 自动显示百分比，startangle: 起始角度
axes2.set_title('世界人口比例',fontsize=20)
#设置图例
#loc: 图例位置，ncol: 图例列数，bbox_to_anchor: 图例位置偏移量（0.5控制水平方位，-0.2控制垂直方位），fontsize: 图例字体大小
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.2),fontsize=12)

#保存图片
plt.savefig('./data/01.png')
plt.show()
```
