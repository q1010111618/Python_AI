```python
import pandas as pd

df=pd.read_csv('./data/sales.csv')

# 2. 数据排序
# 2.1 根据销售数量 倒序排序
df.sort_values('销售数量',ascending=False)

# 2.2 根据 单价 升序排序
df.sort_values('单价')

# 2.3 根据 单价 升序排序，价格一样，再根据 销售数量 降序排序
df.sort_values(['单价','销售数量'],ascending=[True,False])
#df.sort_values(['单价','销售数量'],ascending=True) #如果都是升序，可以直接一个True
```

```python
df= pd.read_csv('data/sales.csv',nrows=30)


#3.分组
#3.1 根据 产品类型 分组，统计各个类型的 订单数量 - count
# 所选列如果有缺失值，会被过滤掉
df.groupby('产品类型')['订单号'].count() 

#3.2 根据 产品类型 分组，统计各个类型的 销售之和 - sum
df.groupby('产品类型')['销售数量'].sum()

#3.3 根据 产品类型 分组，统计各个类型的 销售金额 - sum
df['销售金额']=df['销售数量']*df['单价']
#df
df.groupby('产品类型')['销售金额'].sum()

#3.4 根据 产品类型 分组，统计各个类型的最低商品价格 - min
df.groupby('产品类型')['单价'].min()

#3.5 根据 产品类型 分组，统计各个类型的最高商品价格 - max
df.groupby('产品类型')['单价'].max()

#3.6 根据 产品类型 分组，统计各个类型的平均商品价格 - mean
df.groupby('产品类型')['单价'].mean().round(2)

#3.7 根据 产品类型 分组，统计各个类别的商品的 平均单价，最高单价，最低单价
df.groupby('产品类型')['单价'].agg(['mean','max','min'])

#3.8 根据 产品类型 分组，统计各个类型的 销售数量 之和，销售金额 之和，平均 单价
# 平均单价 保留2位小数
df.groupby('产品类型').agg({'销售数量':'sum','销售金额':'sum','单价':'mean'}).round(2) #或.routd({'单价':2})指定列保留两位小数
```
