```python
df3=pd.read_csv('./data/movie_top250.csv')

# 配置项，展示所有的数据
#pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)

#数据过滤--->df[filter]
# 获取电影分数大于85的电影
df3[df3['电影评分']>85]
#print(df4)

# 获取电影语言为英语或日语的电影
df3[df3['电影语言'].isin(['英语','日语'])]

# 获取电影评分在80到85之间的电影  between-->包含80和85
df3[df3['电影评分'].between(80,85)]

# 获取电影语言为英语且评分大于85的电影
df3[(df3['电影语言']=='英语')&(df3['电影评分']>85)]

# 获取电影语言为英语或日语且电影发布年份2025或2026的电影
df3[(df3['电影语言'].isin(['英语','日语']))&(df3['电影发布年份'].isin(['(2025)','(2026)']))]

#print(df5)
```
