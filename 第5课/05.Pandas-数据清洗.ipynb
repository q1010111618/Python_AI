```python
import pandas as pd
pd.set_option('display.min_rows', 30) #设置显示行数

#1. 读取数据
df = pd.read_csv('./data/movie_top250.csv')

#2. 数据清洗
#2.1 缺失值处理
#df.isnull() #查看缺失值

#2.1.1 删除缺失值
#df.dropna()  #删除缺失值所在行
#df=df.dropna(axis=1) #删除缺失值所在列
#df

#2.1.2 填充缺失值
#df.fillna('未知') #填充缺失值为'未知'
#df.ffill() #填充缺失值，填充上一行数据
#df.bfill() #填充缺失值，填充下一行数据

#2.2 重复值处理
#2.2.1 查看重复值
df.duplicated() #查看重复值（默认检查所有列）
df.duplicated(subset=['电影上映时间']) #查看重复值（只检查指定列）

#2.2.2 删除重复值
df.drop_duplicates(subset=['电影上映时间'],keep='last') #删除重复值,keep='last'保留最后一个重复值,keep='first'保留第一个重复值,keep=False删除所有重复值
```

```python
#2.3 异常值处理
#2.3.1 查看异常值
df[df['电影发布年份']=='(2025)']

#2.3.2 删除异常值
df.drop(df[df['电影发布年份']=='(2025)'].index)

#2.3.3 替换异常值
df['电影发布年份'].replace('(2025)',2025,inplace=True)
```
